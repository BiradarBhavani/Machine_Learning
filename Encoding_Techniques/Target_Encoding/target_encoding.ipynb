# Target Encoding

## 1. Imports

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from category_encoders import TargetEncoder
```

## 2. Load data

```python
df = pd.read_csv('data/loan_data.csv')
df.head()
```

## 3. Define X and y, then split

```python
x = df.drop(columns='loan_status')
y = df['loan_status']
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size=0.8, random_state=42)
```

## 4. Apply Target Encoding

Replaces each category with the mean of the target for that category.
Best for **high-cardinality** nominal features (too many for one-hot).
Fit on train only to avoid leaking target information into the model.
Requires `pip install category_encoders`.

```python
target_encoder = TargetEncoder(cols=['loan_intent'])
te_train = target_encoder.fit_transform(xtrain[['loan_intent']], ytrain)   # needs y, fit on TRAIN
te_test = target_encoder.transform(xtest[['loan_intent']])                 # transform only on TEST

te_train.head()
```
