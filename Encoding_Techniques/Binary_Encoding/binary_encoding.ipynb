# Binary Encoding

## 1. Imports

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from category_encoders import BinaryEncoder
```

## 2. Load data

```python
df = pd.read_csv('data/loan_data.csv')
df.head()
```

## 3. Define X and y, then split

```python
x = df.drop(columns='loan_status')
y = df['loan_status']
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size=0.8, random_state=42)
```

## 4. Apply Binary Encoding

Converts categories to integers, then to binary digits split across
several columns (e.g. 8 categories -> 3 binary columns instead of 8
one-hot columns). Good middle ground for **medium-to-high cardinality**
nominal features — far fewer columns than One-Hot, without the target
leakage risk of Target Encoding. Requires `pip install category_encoders`.

```python
binary_encoder = BinaryEncoder(cols=['loan_intent'])
be_train = binary_encoder.fit_transform(xtrain[['loan_intent']])   # fit_transform on TRAIN
be_test = binary_encoder.transform(xtest[['loan_intent']])          # transform only on TEST

be_train.head()
```
