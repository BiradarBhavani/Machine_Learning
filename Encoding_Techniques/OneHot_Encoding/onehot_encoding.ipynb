# One-Hot Encoding

## 1. Imports

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
```

## 2. Load data

```python
df = pd.read_csv('data/loan_data.csv')
df.head()
```

## 3. Define X and y, then split

```python
x = df.drop(columns='loan_status')
y = df['loan_status']
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size=0.8, random_state=42)
```

## 4. Apply One-Hot Encoding

Creates a new binary (0/1) column per category. Best for **nominal**
(unordered) features with low cardinality. `handle_unknown='ignore'`
avoids errors on unseen categories at test time.

```python
nominal_cols = ['person_gender', 'person_home_ownership', 'loan_intent', 'previous_loan_defaults_on_file']

onehot = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
onehot_train = onehot.fit_transform(xtrain[nominal_cols])   # fit_transform on TRAIN
onehot_test = onehot.transform(xtest[nominal_cols])          # transform only on TEST

onehot_cols = onehot.get_feature_names_out()
pd.DataFrame(onehot_train, columns=onehot_cols).head()
```
