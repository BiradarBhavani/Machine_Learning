# Label Encoding

## 1. Imports

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

## 2. Load data

```python
df = pd.read_csv('data/loan_data.csv')
df.head()
```

## 3. Define X and y, then split

```python
x = df.drop(columns='loan_status')
y = df['loan_status']
xtrain, xtest, ytrain, ytest = train_test_split(x, y, train_size=0.8, random_state=42)
```

## 4. Apply Label Encoding

Converts each category to a single integer. Works on **one column at a
time**. Best for the target column, or a binary categorical feature —
avoid it on nominal features with 3+ categories (it wrongly implies an
order).

```python
label_encoder = LabelEncoder()
encoded_train = label_encoder.fit_transform(xtrain['person_gender'])   # fit_transform on TRAIN
encoded_test = label_encoder.transform(xtest['person_gender'])          # transform only on TEST

print(label_encoder.classes_)
encoded_train[:10]
```

## 5. Put the result back into a DataFrame

```python
pd.DataFrame({'person_gender': xtrain['person_gender'].values, 'encoded': encoded_train}).head()
```
